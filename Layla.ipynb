Jika baris tersebut berisi angka ganjil, tampilkan perkalian angka tersebut dengan dua
Jika Genap, tampilkan apa adanya

```python
with open ('input.txt', 'r') as f:
  for line in f:
    num = int(line.strip())
    if num % 2 != 0 : # if number is odd
        print(num * 2)
    else: # if number is even
        print(num)
```
